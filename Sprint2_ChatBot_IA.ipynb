```python
print("Bem-vindo à Central de Atendimento do seu eletroposto! 🙂")
print("O chatbot Weely estará à disposição para esclarecer suas dúvidas gerais sobre a gestão do seu eletroposto.")
print(f"Digite sua dúvida abaixo:\n")

while True:
    pergunta = input("Você: ").lower()

    msgs.append({'role': 'user', 'content': pergunta})

    resposta =  client.chat.completions.create(
      model="gpt-4o-mini",
      messages=msgs
    )
    conteudo_resposta = resposta.choices[0].message.content
    print()
    print(conteudo_resposta)

    msgs.append({'role': 'assistant', 'content': conteudo_resposta})

    if pergunta == 'sair':
      break
    else:
      print()
      print('Escreva sua pergunta ou escreva "sair" para encerrar o atendimento.')
      print()
```
